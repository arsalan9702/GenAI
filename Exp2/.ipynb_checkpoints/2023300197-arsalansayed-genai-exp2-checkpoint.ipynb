```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.datasets import mnist

print("TF version:", tf.__version__)
```

```python
(x_train, _), (x_test, _) = mnist.load_data()

x_train = x_train.astype("float32")/255.0
x_test = x_test.astype("float32")/255.0

print("Train shape: ", x_train.shape)
print("Test shape: ", x_test.shape)
x_train = x_train.reshape(len(x_train), 784)
x_test = x_test.reshape(len(x_test), 784)

print("Train reshaped: ", x_train.shape)
print("Test reshaped: ", x_test.shape)
```

```python
input_dim = 784
latent_dim = 64
# latent_dim = 128
# latent_dim = 1024

shallow_input = layers.Input(shape=(input_dim,))
encoded = layers.Dense(latent_dim, activation="relu")(shallow_input)
decoded = layers.Dense(input_dim, activation="sigmoid")(encoded)

shallow_ae = models.Model(shallow_input, decoded, name="shallow_autencoder")
shallow_ae.compile(optimizer="adam", loss="binary_crossentropy")
shallow_ae.summary()
```

```python
deep_input = layers.Input(shape=(input_dim,))

x = layers.Dense(256, activation="relu")(deep_input)
x = layers.Dense(128, activation="relu")(x)
encoded_deep = layers.Dense(64, activation="relu")(x)


x = layers.Dense(128, activation="relu")(encoded_deep)
x = layers.Dense(256, activation="relu")(x)
decoded_deep = layers.Dense(input_dim, activation="sigmoid")(x)


deep_ae = models.Model(deep_input, decoded_deep, name="deep_autoencoder")
deep_ae.compile(optimizer="adam", loss="binary_crossentropy")
deep_ae.summary()
```

```python
from tensorflow.keras import backend as K
```

```python
input_dim=784
latent_dim=64

vae_input = layers.Input(shape=(input_dim,))
x = layers.Dense(256, activation="relu")(vae_input)
x = layers.Dense(128, activation="relu")(x)

z_mean = layers.Dense(latent_dim, name="z_mean")(x)
z_log_var = layers.Dense(latent_dim, name="z_log_var")(x)

def sampling(args):
    z_mean, z_log_var = args
    epsilon = K.random_normal(shape=(K.shape(z_mean)[0], latent_dim))
    return z_mean + K.exp(0.5 * z_log_var) * epsilon


z = layers.Lambda(sampling, output_shape=(latent_dim,), name="z")([z_mean, z_log_var])

encoder = models.Model(vae_input, [z_mean, z_log_var, z], name="encoder")
encoder.summary()
```

```python
decoder_input = layers.Input(shape=(latent_dim,))
x = layers.Dense(128, activation="relu")(decoder_input)
x = layers.Dense(256, activation="relu")(x)
decoder_output = layers.Dense(input_dim, activation="sigmoid")(x)

decoder = models.Model(decoder_input, decoder_output, name="decoder")
decoder.summary()
```

```python
class VAE(models.Model):
    def __init__(self, encoder, decoder, **kwargs):
        super().__init__(**kwargs)
        self.encoder = encoder
        self.decoder = decoder
        self.total_loss_tracker = tf.keras.metrics.Mean(name="total_loss")
        self.recon_loss_tracker = tf.keras.metrics.Mean(name="reconstruction_loss")
        self.kl_loss_tracker = tf.keras.metrics.Mean(name="kl_loss")

    @property
    def metrics(self):
        return [self.total_loss_tracker, self.recon_loss_tracker, self.kl_loss_tracker]

    def train_step(self, data):
        x = data
        with tf.GradientTape() as tape:
            z_mean, z_log_var, z = self.encoder(x)
            reconstruction = self.decoder(z)

            reconstruction_loss = tf.reduce_mean(
                tf.reduce_sum(
                    tf.keras.losses.binary_crossentropy(x, reconstruction), axis=0
                )
            ) * input_dim / input_dim  # kept explicit for clarity

            reconstruction_loss = tf.reduce_mean(
                tf.reduce_sum(tf.keras.losses.binary_crossentropy(x, reconstruction), axis=-1)
            )

            kl_loss = -0.5 * tf.reduce_mean(
                tf.reduce_sum(1 + z_log_var - tf.square(z_mean) - tf.exp(z_log_var), axis=-1)
            )

            total_loss = reconstruction_loss + kl_loss

        grads = tape.gradient(total_loss, self.trainable_weights)
        self.optimizer.apply_gradients(zip(grads, self.trainable_weights))

        self.total_loss_tracker.update_state(total_loss)
        self.recon_loss_tracker.update_state(reconstruction_loss)
        self.kl_loss_tracker.update_state(kl_loss)
        return {
            "loss": self.total_loss_tracker.result(),
            "reconstruction_loss": self.recon_loss_tracker.result(),
            "kl_loss": self.kl_loss_tracker.result(),
        }

vae = VAE(encoder, decoder)
vae.compile(optimizer="adam")
```

```python
vae.fit(x_train, epochs=10, batch_size=128, verbose=1)
```

```python
def sampling(args):
    z_mean, z_log_var = args
    dim = tf.shape(z_mean)[1]
    epsilon = K.random_normal(shape=(tf.shape(z_mean)[0], dim))
    return z_mean + tf.exp(0.5 * z_log_var) * epsilon
```

```python
def build_vae_parts(latent_dim, input_dim=784):
    # Encoder
    enc_input = layers.Input(shape=(input_dim,))
    x = layers.Dense(256, activation="relu")(enc_input)
    x = layers.Dense(128, activation="relu")(x)
    z_mean = layers.Dense(latent_dim, name="z_mean")(x)
    z_log_var = layers.Dense(latent_dim, name="z_log_var")(x)
    z = layers.Lambda(sampling, output_shape=(latent_dim,))([z_mean, z_log_var])
    encoder = models.Model(enc_input, [z_mean, z_log_var, z], name="encoder")

    # Decoder
    dec_input = layers.Input(shape=(latent_dim,))
    x = layers.Dense(128, activation="relu")(dec_input)
    x = layers.Dense(256, activation="relu")(x)
    dec_output = layers.Dense(input_dim, activation="sigmoid")(x)
    decoder = models.Model(dec_input, dec_output, name="decoder")

    return encoder, decoder
```

```python
class BetaVAE(models.Model):
    def __init__(self, encoder, decoder, beta=1.0, **kwargs):
        super().__init__(**kwargs)
        self.encoder = encoder
        self.decoder = decoder
        self.beta = beta
        self.total_loss_tracker = tf.keras.metrics.Mean(name="total_loss")
        self.recon_loss_tracker = tf.keras.metrics.Mean(name="reconstruction_loss")
        self.kl_loss_tracker = tf.keras.metrics.Mean(name="kl_loss")

    @property
    def metrics(self):
        return [self.total_loss_tracker, self.recon_loss_tracker, self.kl_loss_tracker]

    def train_step(self, data):
        x = data
        with tf.GradientTape() as tape:
            z_mean, z_log_var, z = self.encoder(x)
            reconstruction = self.decoder(z)

            reconstruction_loss = tf.reduce_mean(
                tf.reduce_sum(tf.keras.losses.binary_crossentropy(x, reconstruction), axis=-1)
            )
            kl_loss = -0.5 * tf.reduce_mean(
                tf.reduce_sum(1 + z_log_var - tf.square(z_mean) - tf.exp(z_log_var), axis=-1)
            )
            total_loss = reconstruction_loss + self.beta * kl_loss

        grads = tape.gradient(total_loss, self.trainable_weights)
        self.optimizer.apply_gradients(zip(grads, self.trainable_weights))

        self.total_loss_tracker.update_state(total_loss)
        self.recon_loss_tracker.update_state(reconstruction_loss)
        self.kl_loss_tracker.update_state(kl_loss)
        return {
            "loss": self.total_loss_tracker.result(),
            "reconstruction_loss": self.recon_loss_tracker.result(),
            "kl_loss": self.kl_loss_tracker.result(),
        }
```

```python
LATENT_DIM = 128   
BETA = 0.5

encoder_b, decoder_b = build_vae_parts(latent_dim=LATENT_DIM)
beta_vae = BetaVAE(encoder_b, decoder_b, beta=BETA)
beta_vae.compile(optimizer="adam")
beta_vae.summary()

history = beta_vae.fit(x_train, epochs=10, batch_size=128, verbose=1)
```

```python
n = 8
_, _, z_test = encoder_b.predict(x_test[:n])
recon = decoder_b.predict(z_test)

plt.figure(figsize=(16, 4))
for i in range(n):
    ax = plt.subplot(2, n, i + 1)
    plt.imshow(x_test[i].reshape(28, 28), cmap="gray")
    plt.title("Original")
    plt.axis("off")

    ax = plt.subplot(2, n, i + 1 + n)
    plt.imshow(recon[i].reshape(28, 28), cmap="gray")
    plt.title(f"beta={BETA}, dim={LATENT_DIM}")
    plt.axis("off")

plt.tight_layout()
plt.show()
```
