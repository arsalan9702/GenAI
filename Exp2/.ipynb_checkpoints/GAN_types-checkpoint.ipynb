```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential, Model
from tensorflow.keras.layers import (
    Dense, Reshape, Flatten, LeakyReLU, BatchNormalization, Dropout,
    Input, Concatenate, Conv2D, Conv2DTranspose, Embedding
)
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.datasets import mnist
import matplotlib.pyplot as plt

(X_train, y_train), (_, _) = mnist.load_data()
X_train = (X_train.astype('float32') - 127.5) / 127.5  # normalize to [-1,1]
X_train = np.expand_dims(X_train, axis=-1)              # (60000,28,28,1)
y_train = y_train.astype('int32')

NOISE_DIM = 100
NUM_CLASSES = 10
IMG_SHAPE = (28, 28, 1)
BATCH_SIZE = 128
EPOCHS = 15000  # adjust as needed

print(X_train.shape, y_train.shape)
```

```python
def plot_generated_images(generator, epoch, noise_dim=NOISE_DIM, labels=None, n=10):
    if labels is None:
        noise = np.random.normal(0, 1, (n, noise_dim))
        gen_imgs = generator.predict(noise, verbose=0)
    else:
        noise = np.random.normal(0, 1, (n, noise_dim))
        gen_imgs = generator.predict([noise, labels], verbose=0)

    gen_imgs = 0.5 * gen_imgs + 0.5  # rescale to [0,1]

    fig, axs = plt.subplots(1, n, figsize=(n, 1.2))
    for i in range(n):
        axs[i].imshow(gen_imgs[i].squeeze(), cmap='gray')
        axs[i].axis('off')
        if labels is not None:
            axs[i].set_title(str(labels[i]), fontsize=8)
    plt.suptitle(f"Epoch {epoch}")
    plt.tight_layout()
    plt.show()
```

```python
def build_generator():
    model = Sequential([
        Dense(256, input_dim=NOISE_DIM),
        LeakyReLU(0.2),
        Dense(512),
        BatchNormalization(),
        LeakyReLU(0.2),
        Dense(1024),
        BatchNormalization(),
        LeakyReLU(0.2),
        Dense(784, activation='tanh'),
        Reshape((28, 28, 1))
    ])
    return model

def build_discriminator():
    model = Sequential([
        Flatten(input_shape=(28, 28, 1)),
        Dense(512),
        LeakyReLU(0.2),
        Dense(256),
        LeakyReLU(0.2),
        Dense(1, activation='sigmoid')
    ])
    return model
```

```python
generator = build_generator()
discriminator = build_discriminator()

g_optimizer = Adam(1e-4, beta_1=0.5)
d_optimizer = Adam(1e-4, beta_1=0.5)
bce = tf.keras.losses.BinaryCrossentropy()

@tf.function
def train_step(real_imgs):
    batch_size = tf.shape(real_imgs)[0]
    noise = tf.random.normal([batch_size, NOISE_DIM])

    with tf.GradientTape() as d_tape, tf.GradientTape() as g_tape:
        gen_imgs = generator(noise, training=True)

        real_output = discriminator(real_imgs, training=True)
        fake_output = discriminator(gen_imgs, training=True)

        # one-sided label smoothing on real labels helps stability
        d_loss_real = bce(tf.ones_like(real_output) * 0.9, real_output)
        d_loss_fake = bce(tf.zeros_like(fake_output), fake_output)
        d_loss = d_loss_real + d_loss_fake

        g_loss = bce(tf.ones_like(fake_output), fake_output)

    d_grads = d_tape.gradient(d_loss, discriminator.trainable_variables)
    g_grads = g_tape.gradient(g_loss, generator.trainable_variables)

    d_optimizer.apply_gradients(zip(d_grads, discriminator.trainable_variables))
    g_optimizer.apply_gradients(zip(g_grads, generator.trainable_variables))

    return d_loss, g_loss

def train_baseline_gan(epochs=EPOCHS, batch_size=BATCH_SIZE, sample_interval=1000):
    for epoch in range(1, epochs + 1):
        idx = np.random.randint(0, X_train.shape[0], batch_size)
        real_imgs = X_train[idx]
        d_loss, g_loss = train_step(real_imgs)

        if epoch % sample_interval == 0:
            print(f"Epoch {epoch} | D loss: {d_loss:.4f} | G loss: {g_loss:.4f}")
            plot_generated_images(generator, epoch)

train_baseline_gan()
```

```python
def build_cgan_generator():
    noise = Input(shape=(NOISE_DIM,))
    label = Input(shape=(1,), dtype='int32')
    label_embedding = Flatten()(Embedding(NUM_CLASSES, NOISE_DIM)(label))

    x = Concatenate()([noise, label_embedding])
    x = Dense(256)(x)
    x = LeakyReLU(0.2)(x)
    x = Dense(512)(x)
    x = BatchNormalization()(x)
    x = LeakyReLU(0.2)(x)
    x = Dense(1024)(x)
    x = BatchNormalization()(x)
    x = LeakyReLU(0.2)(x)
    x = Dense(784, activation='tanh')(x)
    x = Reshape((28, 28, 1))(x)

    return Model([noise, label], x, name='cgan_generator')

def build_cgan_discriminator():
    img = Input(shape=(28, 28, 1))
    label = Input(shape=(1,), dtype='int32')

    label_embedding = Flatten()(Embedding(NUM_CLASSES, 28 * 28)(label))
    label_embedding = Reshape((28, 28, 1))(label_embedding)

    flat_img = Flatten()(img)
    flat_label = Flatten()(label_embedding)
    x = Concatenate()([flat_img, flat_label])
    x = Dense(512)(x)
    x = LeakyReLU(0.2)(x)
    x = Dense(256)(x)
    x = LeakyReLU(0.2)(x)
    x = Dense(1, activation='sigmoid')(x)

    return Model([img, label], x, name='cgan_discriminator')
```

```python
cgan_generator = build_cgan_generator()
cgan_discriminator = build_cgan_discriminator()

g_optimizer = Adam(1e-4, beta_1=0.5)
d_optimizer = Adam(1e-4, beta_1=0.5)
bce = tf.keras.losses.BinaryCrossentropy()

@tf.function
def train_step_cgan(real_imgs, real_labels):
    batch_size = tf.shape(real_imgs)[0]
    noise = tf.random.normal([batch_size, NOISE_DIM])
    fake_labels = tf.random.uniform([batch_size, 1], 0, NUM_CLASSES, dtype=tf.int32)

    with tf.GradientTape() as d_tape, tf.GradientTape() as g_tape:
        gen_imgs = cgan_generator([noise, fake_labels], training=True)

        real_output = cgan_discriminator([real_imgs, real_labels], training=True)
        fake_output = cgan_discriminator([gen_imgs, fake_labels], training=True)

        d_loss_real = bce(tf.ones_like(real_output) * 0.9, real_output)
        d_loss_fake = bce(tf.zeros_like(fake_output), fake_output)
        d_loss = d_loss_real + d_loss_fake

        g_loss = bce(tf.ones_like(fake_output), fake_output)

    d_grads = d_tape.gradient(d_loss, cgan_discriminator.trainable_variables)
    g_grads = g_tape.gradient(g_loss, cgan_generator.trainable_variables)

    d_optimizer.apply_gradients(zip(d_grads, cgan_discriminator.trainable_variables))
    g_optimizer.apply_gradients(zip(g_grads, cgan_generator.trainable_variables))

    return d_loss, g_loss

def train_cgan(epochs=EPOCHS, batch_size=BATCH_SIZE, sample_interval=1000):
    for epoch in range(1, epochs + 1):
        idx = np.random.randint(0, X_train.shape[0], batch_size)
        real_imgs = X_train[idx]
        real_labels = y_train[idx].reshape(-1, 1).astype(np.int32)

        d_loss, g_loss = train_step_cgan(real_imgs, real_labels)

        if epoch % sample_interval == 0:
            print(f"Epoch {epoch} | D loss: {d_loss:.4f} | G loss: {g_loss:.4f}")
            plot_generated_images(cgan_generator, epoch, labels=np.arange(10).reshape(-1, 1))

train_cgan()
```

```python
def generate_digit(digit, n_samples=5):
    noise = np.random.normal(0, 1, (n_samples, NOISE_DIM))
    labels = np.full((n_samples, 1), digit)
    gen_imgs = cgan_generator.predict([noise, labels], verbose=0)
    gen_imgs = 0.5 * gen_imgs + 0.5

    fig, axs = plt.subplots(1, n_samples, figsize=(n_samples, 1.5))
    for i in range(n_samples):
        axs[i].imshow(gen_imgs[i].squeeze(), cmap='gray')
        axs[i].axis('off')
    plt.suptitle(f"Generated digit: {digit}")
    plt.show()

for i in range(10):
  generate_digit(i)

```

```python
def build_cdcgan_generator():
    noise = Input(shape=(NOISE_DIM,))
    label = Input(shape=(1,), dtype='int32')

    label_embedding = Flatten()(Embedding(NUM_CLASSES, NOISE_DIM)(label))
    x = Concatenate()([noise, label_embedding])  # (batch, NOISE_DIM*2)

    x = Dense(7*7*256)(x)
    x = BatchNormalization()(x)
    x = LeakyReLU(0.2)(x)
    x = Reshape((7, 7, 256))(x)

    x = Conv2DTranspose(128, 4, strides=2, padding='same')(x)
    x = BatchNormalization()(x)
    x = LeakyReLU(0.2)(x)

    x = Conv2DTranspose(64, 4, strides=2, padding='same')(x)
    x = BatchNormalization()(x)
    x = LeakyReLU(0.2)(x)

    x = Conv2D(1, 7, padding='same', activation='tanh')(x)

    return Model([noise, label], x, name='cdcgan_generator')


def build_cdcgan_discriminator():
    img = Input(shape=IMG_SHAPE)
    label = Input(shape=(1,), dtype='int32')

    # Project label embedding into a 28x28x1 feature map, concat as extra channel
    label_embedding = Embedding(NUM_CLASSES, 28 * 28)(label)
    label_embedding = Flatten()(label_embedding)
    label_map = Reshape((28, 28, 1))(label_embedding)

    x = Concatenate(axis=-1)([img, label_map])  # (28,28,2)

    x = Conv2D(64, 4, strides=2, padding='same')(x)
    x = LeakyReLU(0.2)(x)
    x = Dropout(0.3)(x)

    x = Conv2D(128, 4, strides=2, padding='same')(x)
    x = LeakyReLU(0.2)(x)
    x = Dropout(0.3)(x)

    x = Flatten()(x)
    x = Dense(1, activation='sigmoid')(x)

    return Model([img, label], x, name='cdcgan_discriminator')
```

```python
cdcgan_generator = build_cdcgan_generator()
cdcgan_discriminator = build_cdcgan_discriminator()

g_optimizer = Adam(2e-4, beta_1=0.5)
d_optimizer = Adam(5e-5, beta_1=0.5)  # slightly slower D, per earlier balancing discussion
bce = tf.keras.losses.BinaryCrossentropy()

@tf.function
def train_step_cdcgan(real_imgs, real_labels):
    batch_size = tf.shape(real_imgs)[0]
    noise = tf.random.normal([batch_size, NOISE_DIM])
    fake_labels = tf.random.uniform([batch_size, 1], 0, NUM_CLASSES, dtype=tf.int32)

    with tf.GradientTape() as d_tape, tf.GradientTape() as g_tape:
        gen_imgs = cdcgan_generator([noise, fake_labels], training=True)

        real_output = cdcgan_discriminator([real_imgs, real_labels], training=True)
        fake_output = cdcgan_discriminator([gen_imgs, fake_labels], training=True)

        d_loss_real = bce(tf.ones_like(real_output) * 0.9, real_output)
        d_loss_fake = bce(tf.zeros_like(fake_output), fake_output)
        d_loss = d_loss_real + d_loss_fake

        g_loss = bce(tf.ones_like(fake_output), fake_output)

    d_grads = d_tape.gradient(d_loss, cdcgan_discriminator.trainable_variables)
    g_grads = g_tape.gradient(g_loss, cdcgan_generator.trainable_variables)

    d_optimizer.apply_gradients(zip(d_grads, cdcgan_discriminator.trainable_variables))
    g_optimizer.apply_gradients(zip(g_grads, cdcgan_generator.trainable_variables))

    return d_loss, g_loss
```

```python
d_losses_cdcgan, g_losses_cdcgan = [], []

def train_cdcgan(epochs=EPOCHS, batch_size=BATCH_SIZE, sample_interval=1000):
    for epoch in range(1, epochs + 1):
        idx = np.random.randint(0, X_train.shape[0], batch_size)
        real_imgs = X_train[idx]
        real_labels = y_train[idx].reshape(-1, 1).astype(np.int32)

        d_loss, g_loss = train_step_cdcgan(real_imgs, real_labels)
        d_losses_cdcgan.append(float(d_loss))
        g_losses_cdcgan.append(float(g_loss))

        if epoch % sample_interval == 0:
            print(f"Epoch {epoch} | D loss: {d_loss:.4f} | G loss: {g_loss:.4f}")
            plot_generated_images(cdcgan_generator, epoch, labels=np.arange(10).reshape(-1, 1))

train_cdcgan()
```

```python
def generate_digit_dcgan(digit, n_samples=5):
    noise = np.random.normal(0, 1, (n_samples, NOISE_DIM))
    labels = np.full((n_samples, 1), digit)
    gen_imgs = cdcgan_generator.predict([noise, labels], verbose=0)
    gen_imgs = 0.5 * gen_imgs + 0.5

    fig, axs = plt.subplots(1, n_samples, figsize=(n_samples, 1.5))
    for i in range(n_samples):
        axs[i].imshow(gen_imgs[i].squeeze(), cmap='gray')
        axs[i].axis('off')
    plt.suptitle(f"Generated digit: {digit}")
    plt.show()

for i in range(10):
  generate_digit_dcgan(i)
```

```python
def print_summary(model, name):
    print(name)
    model.summary()
    print()

# Baseline GAN
print_summary(generator, "BASELINE GENERATOR")
print_summary(discriminator, "BASELINE DISCRIMINATOR")

# Conditional GAN
print_summary(cgan_generator, "CGAN GENERATOR")
print_summary(cgan_discriminator, "CGAN DISCRIMINATOR")

# DCGAN
print_summary(cdcgan_generator, "DCGAN GENERATOR")
print_summary(cdcgan_discriminator, "DCGAN DISCRIMINATOR")
```
